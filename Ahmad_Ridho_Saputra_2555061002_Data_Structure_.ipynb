Norms and Unit Vectors

```python
# Membuat Vector x dengan ukuran 3x1
x = np.array([20, 2, 55])
x
```

```python
# Menghitung L2 Norm dari vector x
(20**2 + 2**2 + 55**2)**(1/2)
```

```python
np.linalg.norm(x)
```

```python
# 2. buat vector v dengan elemen (3,4)
#hitung nilai L2 norm dari vector v
#tampilkan unit vector dari vector v
#hitung panjang unit vector dari vector v

# Membuat vektor
v = np.array([3, 4])

# Menghitung panjang (norma L2)
norm = np.linalg.norm(v)

# Menghitung unit vector
unit_v = v / norm

print("Vektor asli:", v)
print("Panjang vektor:", norm)
print("Unit vector:", unit_v)
print("Panjang unit vector:", np.linalg.norm(unit_v))
```

Basis, Orthogonal and Orthonormal Vectors

```python
import numpy as np
```

```python
# buat vector i dengan elemen (1,0) dan vector j dengan elemen (0,1)
i = np.array([1, 0])
i
```

```python
j = np.array([0, 1])
j
```

```python
# Membuktikan bahwa vector i dan j orthogonal
np.dot(i, j)
print (np.dot (i, j))

# Jika hasilnya adalah 0 maka nilainya orthogonal
```

```python
# buat vector v dengan elemen (1,2) dan vector w dengan elemen (2,-1)
v = np.array([1, 2])
v
```

```python
w = np.array([2, -1])
w
```

```python
# Membuktikkan apakah vector v dan w orthogonal
np.dot(v, w)
print (np.dot (v, w))
# # Jika hasilnya adalah 0 maka nilainya orthogonal
```
